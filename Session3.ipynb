Charts For Numerical Data

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.bar(data['bedrooms'].value_counts().index, data['bedrooms'].value_counts())
plt.show()
```

```python
sns.countplot(x='bedrooms',hue='furnishingstatus', data=data)
plt.show()
```

```python
plt.hist(data['price'], bins=70)
plt.show()
```

```python
plt.scatter(data['area'], data['price'], marker='o', color='green')
plt.title('Area vs Price')
plt.xlabel('Area')
plt.ylabel('Price')
plt.show()
```

```python
plt.scatter(data['area'], data['price'], marker='o', c=data['price'])
plt.title('Area vs Price')
plt.xlabel('Area')
plt.ylabel('Price')
plt.show()
```

```python
plt.boxplot(data['price'])  
plt.show()
```

Charts For Categorical Values

```python
furnishing_counts = data['furnishingstatus'].value_counts()
colors = ['blue', 'green', 'red']

bars = plt.bar(furnishing_counts.index, furnishing_counts, color=colors)
plt.xlabel('Furnishing Status')
plt.ylabel('Number of Houses')
plt.title('Number of Houses by Furnishing Status')

plt.legend(bars, furnishing_counts.index, title='Furnishing Status')
plt.show()
```

```python
sns.countplot(x='furnishingstatus', data=data)
plt.show()
```

```python
plt.pie(furnishing_counts, labels=furnishing_counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Distribution of Furnishing Status')
plt.show()
```

Catplot in seaborn allow us to plot a categorical variable with a numerical one with different charts

```python
sns.catplot(x='furnishingstatus', y='price', data=data, kind='bar')
plt.show()
```

```python
sns.catplot(x='airconditioning', y='price', data=data, kind='box')
plt.show()
```
