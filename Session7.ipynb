```python
cols = ['sepal length', 'sepal width', 'petal length', 'petal width', 'Flower Type']
data = pd.read_csv('/kaggle/input/datasets/vikrishnan/iris-dataset/iris.data.csv', header = None, names=cols)
data.head()
```

```python
data.isnull().sum()
```

```python
target = 'Flower Type'
X = data.drop(target, axis=1)
y = data[target]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state = 75, stratify=y)
```

```python
numerical_cols = list(X.columns)
numerical_cols
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train[numerical_cols] = scaler.fit_transform(X_train[numerical_cols])
X_test[numerical_cols] = scaler.transform(X_test[numerical_cols])
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
data[target].value_counts()
```

```python
from sklearn.metrics import accuracy_score

y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)

train_acc = accuracy_score(y_train, y_pred_train)
test_acc = accuracy_score(y_test, y_pred_test)

print(f'Train accuracy {train_acc}')
print(f'Test accuracy {test_acc}')
```
