```python
## Ex 2
print("plz enter an even len text")
ev_text = input("ev_text")
if len(ev_text) %2 == 0:
    text2 = ev_text[0::2]

    mid = len(text2) // 2

    text3 = text2[:mid] + text2[mid:][::-1]

    text4 = text3[-1] + text3[1:-1] + text3[0]

    print(text4.upper())

else:
    print("Please enter a text with even length")


```

```python
## EX 3
row, cols = (3,3)
## matrix = [cols*[0]] * row


matrix = [[0]*3 for _ in range(3)]
matrix 

for i in range(3):
   for j in range(3):
      if i == j and i + j == 2:
         matrix[i][j] = 3
      elif i == j:
            matrix[i][j] = 1
      elif i + j == 2:
            matrix[i][j] = 2
        
print(matrix)

for row in matrix:
    print(*row)

```
