```python
sales=np.random.uniform(100, 500, size = (4, 7))
sales[sales > 300] *= 1.10
sales = np.where(sales > 300, sales * 0.95, sales)
weekends = sales[:,-2:]
avg_weekends = np.mean(weekends, axis=1)
avg_weekends
```
